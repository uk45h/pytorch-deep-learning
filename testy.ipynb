```python
import matplotlib.pyplot as plt
import numpy as np

# Dane wejściowe
g = 9.81
h_max = 3.0
h_target = 0.5

# Obliczenia
v0 = np.sqrt(2 * g * h_max)
t_up = v0 / g
t_total = 2 * t_up

t = np.linspace(0, t_total, 500)
v = v0 - g * t

# Czas dotarcia do 0.5m (dwa rozwiązania równania kwadratowego)
# h = v0*t - 0.5*g*t^2  => 0.5*g*t^2 - v0*t + h = 0
delta = v0**2 - 2*g*h_target
t_05_up = (v0 - np.sqrt(delta)) / g
t_05_down = (v0 + np.sqrt(delta)) / g

v_at_05 = v0 - g * t_05_up

# Wykres
plt.figure(figsize=(10, 6))
plt.plot(t, v, label='Predkosc v(t)', color='blue', linewidth=2)
plt.axhline(0, color='black', linewidth=1)
plt.axvline(t_up, color='red', linestyle='--', label='Punkt zwrotny (3m)')

# Punkty na 0.5m
plt.scatter([t_05_up, t_05_down], [v_at_05, -v_at_05], color='green', zorder=5)

plt.text(t_05_up, v_at_05 + 0.5, f' Wznoszenie: {v_at_05:.2f} m/s')
plt.text(t_05_down, -v_at_05 - 1.5, f' Spadanie: {-v_at_05:.2f} m/s')

plt.title('Predkosc w zaleznosci od czasu (v od t)')
plt.xlabel('Czas [s]')
plt.ylabel('Predkosc [m/s]')
plt.grid(True, linestyle=':', alpha=0.6)
plt.legend()
plt.show()
```
